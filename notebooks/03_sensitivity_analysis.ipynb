# 03. Parameter Sensitivity Analysis

Investigates the sensitivity of predicted path loss to:
- Base station antenna height ($h_t$)
- Mobile antenna height ($h_r$)
- Carrier frequency ($f$)
- Path loss exponent ($n$)


```python
import sys
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))

from propagation.analysis.sensitivity import SensitivityAnalyzer

# 1. Base Station Height Sensitivity Sweep
df_ht = SensitivityAnalyzer.sweep_tx_antenna_height(
    min_ht_m=20.0, max_ht_m=200.0, num_points=50, distance_km=5.0, frequency_mhz=900.0
)

plt.figure(figsize=(7, 4.5))
plt.plot(df_ht['tx_height_m'], df_ht['PL_TwoRay_dB'], 'orange', lw=2, label='Two-Ray Ground Refl.')
plt.plot(df_ht['tx_height_m'], df_ht['PL_Okumura_dB'], 'red', lw=2, ls=':', label='Okumura')
plt.plot(df_ht['tx_height_m'], df_ht['PL_Hata_dB'], 'purple', lw=2, ls='--', label='Okumura-Hata')
plt.xlabel('Base Station Antenna Height ht (m)')
plt.ylabel('Path Loss (dB)')
plt.title('Path Loss Sensitivity to Base Station Antenna Height (d = 5 km)')
plt.grid(True, ls=':')
plt.legend()
plt.show()
```

### 2D Parameter Heatmap: Distance vs Carrier Frequency


```python
D, F, PL = SensitivityAnalyzer.generate_2d_heatmap_grid(
    model_name='cost231',
    distances_km=np.linspace(1.0, 15.0, 50),
    frequencies_mhz=np.linspace(1500.0, 2000.0, 50)
)

plt.figure(figsize=(8, 5.5))
contour = plt.contourf(D, F, PL, levels=25, cmap='plasma')
cbar = plt.colorbar(contour)
cbar.set_label('Path Loss (dB)')
plt.xlabel('Distance (km)')
plt.ylabel('Frequency (MHz)')
plt.title('COST-231 Hata 2D Path Loss Parameter Space')
plt.show()
```
